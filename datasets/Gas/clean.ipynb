# Gas transport dataset cleanup


```python
from pathlib import Path

import pandas as pd
import yaml
```

```python
data_dir = Path.cwd()
config = yaml.safe_load((data_dir / 'dataset.yaml').read_text())
raw_path = data_dir / 'Gas_permeability_solubility_diffusivity_wide.csv'
clean_path = data_dir / config['data_file']
smiles_col = config['smiles_column']

df = pd.read_csv(raw_path, low_memory=False)
df = df.groupby(smiles_col, sort=False, as_index=False).median()

star_count = df[smiles_col].str.count(r'\*')
df = df.loc[star_count.eq(2)].copy()

print(f'Input rows: {star_count.shape[0]}')
print(f'Removed rows by star count: {(~star_count.eq(2)).sum()}')
print(f'Rows after star-count filter: {len(df)}')
```

```python
df.to_csv(clean_path, index=False)
df
```
